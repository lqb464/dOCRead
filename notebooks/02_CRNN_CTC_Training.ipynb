# 02 — Custom CRNN + CTC Loss Training

**Goal**: Train a custom Convolutional Recurrent Neural Network (CRNN) with Connectionist Temporal Classification (CTC) Loss from scratch.

```python
import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

CTC Loss is crucial for OCR because the input image width does not perfectly map 1:1 to the output sequence length. CTC handles this by calculating the probability of all possible alignments and summing them up.

```python
from src.models.crnn import CRNN

# Create CRNN model
model = CRNN(imgH=32, nc=1, nclass=37, nh=256).to(device)

# Output shape will be [Sequence Length, Batch Size, Num Classes]
dummy_input = torch.randn(4, 1, 32, 100).to(device)
out = model(dummy_input)
print("CRNN Output Shape (T, N, C):", out.shape)
```
