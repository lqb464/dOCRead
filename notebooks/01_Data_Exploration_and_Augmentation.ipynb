# 01 — Data Exploration and Augmentation (Synthetic OCR)

**Goal**: Visualize the synthetic datasets generated for Printed and Handwritten text, and explore image augmentations to make the models robust against real-world noise.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw, ImageFont
import random
```

### Creating a Sample Image (Simulating `trdg` output)

```python
def create_text_image(text, is_handwritten=False):
    # Create a white background
    img = Image.new('RGB', (400, 64), color = (255, 255, 255))
    d = ImageDraw.Draw(img)
    
    # In reality, you'd load a TTF font here.
    # For this notebook without external assets, we use default.
    d.text((20, 20), text, fill=(0, 0, 0))
    return np.array(img)

img = create_text_image("This is a test document.")
plt.imshow(img)
plt.title("Clean Text Image")
plt.axis('off')
plt.show()
```

### Applying Augmentations
Models must learn to ignore shadows, blurs, and perspective shifts.

```python
def add_noise(image):
    noise = np.random.normal(0, 15, image.shape).astype(np.uint8)
    noisy = cv2.add(image, noise)
    return noisy

def apply_blur(image):
    return cv2.GaussianBlur(image, (5, 5), 0)

noisy_img = add_noise(img)
blurred_img = apply_blur(noisy_img)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].imshow(noisy_img)
ax[0].set_title("Gaussian Noise")
ax[1].imshow(blurred_img)
ax[1].set_title("Noise + Blur")
plt.show()
```
